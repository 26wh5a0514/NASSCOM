```python
#Second Largest Element
ls=[int(x) for x in input("Enter list of numbers:").split()]
print(ls)
l1=list()

for x in range(len(ls)):
  count=0
  for num in ls:
    if ls[x]==num:
      count+=1
  if count==1:
    l1.append(ls[x])
print(l1)
for x in range(0,len(l1)-1):
  if l1[x]<l1[x+1]:
    l1[x],l1[x+1]=l1[x+1],l1[x]
if len(l1)>=2:
    print("second largest unique number:",l1[1])
else:
    print("no second largest unique number!")

```

```python
#frequency of numbers
l1=[int(x) for x in input("Enter list of numbers:").split()]
dict={}
for x in l1:
  if x in dict:
    dict[x]+=1
  else:
    dict[x]=1
print("frequency of each number:",dict)
```
