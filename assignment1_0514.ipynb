```python
#Second Largest Element
ls=[int(x) for x in input("Enter list of numbers:").split()]
print(ls)
l1=list()

for x in range(len(ls)):
  count=0
  for num in ls:
    if ls[x]==num:
      count+=1
  if count==1:
    l1.append(ls[x])
print(l1)
for x in range(0,len(l1)-1):
  if l1[x]<l1[x+1]:
    l1[x],l1[x+1]=l1[x+1],l1[x]
if len(l1)>=2:
    print("second largest unique number:",l1[1])
else:
    print("no second largest unique number!")

```

```python
#Pair Sum
ls=[int(x) for x in input("Enter list of numbers:").split()]
target=int(input("Enter target value:"))
seen=set()
printed=set()
for x in ls:
  complement=target-x
  if complement in seen:
    pair=tuple(sorted((x,complement)))
    if pair not in printed:
      print(pair)
      printed.add(pair)
  seen.add(x)
```

```python
#Longest consecutive sequence
a = [int(x) for x in input("Enter list of numbers: ").split()]

a.sort()

count = 1
maximum = 1

for i in range(1, len(a)):
    if a[i] == a[i - 1] + 1:
        count += 1
    elif a[i] != a[i - 1]:
        count = 1

    if count > maximum:
        maximum = count

print("Longest consecutive sequence:", maximum)
```

```python
#Matrix Addition, Subtraction and Multiplication
r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))

print("Enter first matrix:")
a = []

for i in range(r):
    a.append([int(x) for x in input().split()])

print("Enter second matrix:")
b = []

for i in range(r):
    b.append([int(x) for x in input().split()])

# Addition
print("Addition:")

for i in range(r):
    for j in range(c):
        print(a[i][j] + b[i][j], end=" ")
    print()

# Subtraction
print("Subtraction:")

for i in range(r):
    for j in range(c):
        print(a[i][j] - b[i][j], end=" ")
    print()

# Multiplication
print("Multiplication:")

for i in range(r):
    for j in range(c):
        total = 0

        for k in range(c):
            total = total + a[i][k] * b[k][j]

        print(total, end=" ")

    print()
```

```python
#Spiral Traversal
n = int(input("Enter size of matrix: "))

a = []

print("Enter matrix:")

for i in range(n):
    a.append([int(x) for x in input().split()])

top = 0
bottom = n - 1
left = 0
right = n - 1

print("Spiral order:")

while top <= bottom and left <= right:

    # Left to right
    for j in range(left, right + 1):
        print(a[top][j], end=" ")

    top = top + 1

    # Top to bottom
    for i in range(top, bottom + 1):
        print(a[i][right], end=" ")

    right = right - 1

    # Right to left
    if top <= bottom:
        for j in range(right, left - 1, -1):
            print(a[bottom][j], end=" ")

        bottom = bottom - 1

    # Bottom to top
    if left <= right:
        for i in range(bottom, top - 1, -1):
            print(a[i][left], end=" ")

        left = left + 1
```

```python
#Majority element
a = [int(x) for x in input("Enter list of numbers: ").split()]
n = len(a)
found = False

for x in a:
    if a.count(x) > n / 2:
        print("Majority Element:", x)
        found = True
        break

if found == False:
    print("No Majority Element")
```

```python
#Stock Buy and Sell
prices = [int(x) for x in input("Enter stock prices: ").split()]
minimum = prices[0]
profit = 0

for price in prices:
    if price < minimum:
        minimum = price

    if price - minimum > profit:
        profit = price - minimum

print("Maximum Profit:", profit)
```
